```python
# Styling and Customization 

# 1.Colormaps(cmap) : numerical values to colors 
#    useful for scatter plots , heatmaps and other visualizations where color rep a value .

import matplotlib.pyplot as plt 
import numpy as np 

x,y=np.random.randn(100),np.random.randn(100)

values=np.random.randn(100)

plt.scatter(x,y,c=values,cmap="coolwarm",s=80) # "viridis", "plasma", "coolwarm", "magma" and "cividis".
plt.colorbar(label="Value")
plt.title("Scatter plot with colormap")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()



```

```python
# Style Sheets  : it changes overall appearnace of plots 
 

plt.style.use("classic")

x=[i for i in range(1,6)]
y=[10,15,13,20,25]

plt.plot(x,y,marker="*",linewidth=2)
plt.title("Styled Line Graph")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.show()

# Other Built-in style : "default", "bmh", "classic" and "dark_background".
print(plt.style.available)
```

```python
# rcParams : global plot settings 

# rcParams lets you customize Matplotlib's def settings , such as font size , figures dim's and lineWidth 



plt.rcParams["figure.figsize"]=(8,5)
plt.rcParams["font.size"]=12
plt.rcParams["lines.linewidth"]=2
plt.rcParams["axes.titlesize"]=16
plt.rcParams["axes.grid"]=True 

x = [1, 2, 3, 4]
y = [10, 25, 20, 35]

plt.plot(x, y, marker="o")
plt.title("Customized Figure")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
# To avoid changing defaults for the rest of your notebook, use a temporary configuration:
with plt.rc_context({"font.size": 14, "axes.grid": True}):
    plt.plot([1, 2, 3], [3, 5, 4])
    plt.title("Temporary Settings")
    plt.show()
```

```python
# Custom Themes : 
# We can combine a style sheet , rcParams , colors and fig to create a reusable theme

plt.style.use("dark_background")

plt.rcParams.update({
    "figure.figsize": (8, 5),
    "font.size": 12,
    "axes.titlesize": 16,
    "lines.linewidth": 3
})

plt.plot(
    [1, 2, 3, 4, 5],
    [10, 20, 15, 30, 25],
    color="cyan",
    marker="o"
)
plt.show()
```

```python
# Spines : Customize plot borders , borders sorrounding the plotting area
#  You can hide, show or reposition them.

fig,ax=plt.subplots()

ax.plot([1,2,3,4],[2,5,3,7],marker="o")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.spines["left"].set_linewidth(2)
ax.spines["bottom"].set_linewidth(2)

ax.set_title('Customized Spines')
plt.show()
# Common spine names are "top", "bottom", "left" and "right".
```

```python
# Annotations : Highlight imp points 
# Annotations add explanatory text and arrows to specific points in a plot.

x=[1,2,3,4,5]
y=[10,15,12,30,20]

fig,ax=plt.subplots() # def : 1

ax.plot(x,y,marker="*")
ax.annotate(
    "Highest value",
    xy=(4,30),
    xytext=(2.5,32),
    arrowprops={"arrowstyle":"->","color":"red"},
    color="yellow"
)

ax.set_title("Sales with Annotations")
ax.set_xlabel("Day")
ax.set_ylabel("Sales")
plt.show()

# xy — loc of point being highlighted
# xytext — pos'n of annotation text
# arrowprops — arrow styling

```
