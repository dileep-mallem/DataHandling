```python
# Subplots -> aloows you to display multiplots plots(barpie,hist etc) inside one figure

# 1. basic gid of subplots 
# Use plt.subplots(rows,cols) to create a regular grid ,
#     It returns tuples of two objects : Figures obj first and axes obj second 

# fig : the Overall container or window frame that hold everything 
#       It manges figsixe, super titles fig.suptitle() and saveing final image fig.savefig()

# ax (axes/subplot) : indivial plot canvas , ax.set_title() ax.set_xlabel() 
import matplotlib.pyplot as plt 
import numpy as np

fig,ax = plt.subplots(2,2,figsize=(10,7)) # canvas, plot_area = plt.subplots()
fig.suptitle("First SubPlot")
x=[i for i in range(1,6)]
expenses=["Food","travel","Books","Others"]
amounts=[3000,1000,1500,500]


ax[0,0].plot(x,[2,4,3,5,7],marker='*',color='red',linestyle=':')
ax[0,0].set_title("Line Graph")

ax[0,1].bar(["A","b","c"],[10,20,15])
ax[0,1].set_title("bar graph")

ax[1,0].pie(amounts,labels=expenses,autopct="%1.1f%%",startangle=90)
ax[1,0].set_title("Pie Chart")

ax[1,1].hist(np.random.randn(100),bins=10)
ax[1,1].set_title("histogram")

plt.tight_layout() # acts like an automatic spacer and layout clean-up tool. 
# Adjusts padding and margins between subplots , sep subplots in thier own boxes , preventing ovelapping of sub titkes and axis labels
plt.show()

# ax[0, 0] means first row, first column.
# figsize controls the overall figure size.
```

```python
# Shared Axes 
# use sharex=true or sharey=True to share an axis across subplots 
# Useful when comparing datasets on the same scale 

import numpy as np 
import matplotlib.pyplot as plt

x=np.arange(1,6)

fig,ax=plt.subplots(2,1,figsize=(7,6),sharex=True)
fig.suptitle("Shared Axes")

ax[0].plot(x,[10,20,15,30,25],color='green',marker='o')
ax[0].set_title("Class A")
ax[0].set_ylabel("Marks")

ax[1].plot(x,[15,25,20,35,30],marker='x')
ax[1].set_title("Class B")
ax[1].set_xlabel("Test Number")
ax[1].set_ylabel("Marks")

plt.tight_layout()
plt.show()
```

```python
# Unequal Subplots layouts 
# Use when one plot needs more space than the others 

import numpy as np 
import matplotlib.pyplot as plt

fig,ax=plt.subplots(2,2,figsize=(10,6),gridspec_kw={"width_ratios":[2,1]})

x=np.arange(1,8)

ax[0, 0].plot(x, x**2, color="blue")
ax[0, 0].set_title("Large Plot")

ax[0, 1].bar(["A", "B"], [5, 8])
ax[0, 1].set_title("Small Plot")

ax[1,0].plot(x,np.sqrt(x),color="green")
ax[1,0].set_title("Large Plot 2")

ax[1,1].scatter([1,2,3],[3,1,4])
ax[1,1].set_title("Small Plot 2")

plt.tight_layout()
plt.show()

# here left col is twice than right one
```

```python
# Gris spec : adv subplot layouts 
# It gives precise control over how rows and cols are divided 
# A subplot can span multiple rows or cols

import numpy as np 
import matplotlib.pyplot as plt

fig=plt.figure(figsize=(10,6))

gs=fig.add_gridspec(
    2,2,
    width_ratios=[2,1],
    height_ratios=[1,1]
)
# width_ratios — controls rel colmn widths.
# height_ratios — controls rel row heights.

ax1=fig.add_subplot(gs[:,0]) # Both rows , left col
ax2=fig.add_subplot(gs[0,1]) # top-right
ax3=fig.add_subplot(gs[1,1]) # bottom right 

x = np.arange(1, 11)

ax1.plot(x, x**2, marker="o")
ax1.set_title("Main Plot")

ax2.bar(["A", "B", "C"], [5, 8, 6])
ax2.set_title("Bar Chart")

ax3.scatter(x, np.random.randint(1, 10, 10))
ax3.set_title("Scatter Plot")

plt.tight_layout()
plt.show()
```
