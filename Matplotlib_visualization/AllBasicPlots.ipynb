```python
import matplotlib.pyplot as plt 

# Line Plot Parameters 
# color : 'red','c0','#FF4e4e' -> Line color
# linewidth/lw : 1.2,2 -> Line Thickness in points 
# linestyle/ls : -,--,:,-. -> Solid,dashed,dottted,dash-dot
# marker : o,s,^,*,x 
# alpha : 0.0-1.0 -> Tansparency 
# label : trail loss -> legend text for this line 
# zorder : 1,2,10 -> Drawing Order , Higher = drawn on top

# Lets Practice Some 

```

```python
import matplotlib.pyplot as plt 
x=[i for i in range(-5,6)]
y=[i**2 for i in range(-5,6)]

plt.plot(x,y) # for simple line graph

plt.xlabel("Numbers")
plt.ylabel("Squares of Numbers")
plt.title("Squares Graph") # Title of Graph 

plt.show() # It returns the resultant plots
```

```python
import matplotlib.pyplot as plt 

# Temp over 4 days 
days=["Mon","tue","Wed","Thurs"]
temp=[28,34,56,31]

plt.plot(days,temp,marker="*",linestyle=":",color="#75b800")
plt.xlabel("Days")
plt.ylabel("temperatures")
plt.title("temp Over 4 Days")
plt.grid(True) # Visible bool ,def = None
plt.show()
```

```python
import matplotlib.pyplot as plt 

# Bar Graph of marks in diff subjects 

subjects=["Python","ML","DS","DSA"]
marks=[70,100,80,60]

plt.bar(subjects,marks) # for bar 
plt.xlabel("Subjects")
plt.ylabel("marks")
plt.title("Subject-Wise Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt 

products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [40, 75, 30, 55]

plt.bar(products,sales,color='orange',width=0.5,align='center')
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Units Sold")

plt.show()
```

```python
import matplotlib.pyplot as plt 


acticvities=["Sleep","Study","College","Leisure"]
hours=[8,7,6,3]

plt.pie(hours,labels=acticvities,colors=['red','blue','orange','green'],autopct="%1.1f%%")

plt.title("Daily Time Allocation")
plt.show()
```

```python
import matplotlib.pyplot as plt 

# labels: Names of the categories.
# autopct: Displays percentages.
# startangle: Rotates the pie chart.
# explode: Separates selected slices from the pie.


expenses=["Food","travel","Books","Others"]
amounts=[3000,1000,1500,500]

plt.pie(amounts,labels=expenses,autopct="%1.1f%%",startangle=60,explode=[0.1,0,0,0],shadow=True,pctdistance=0.5) # pctdist -> % on parts dist from center 

plt.title("Monthly Expenses")

plt.show()
```

```python
import matplotlib.pyplot as plt 


# Scatter Plt : Displays the realationship between two Numerical variables

hours=[1,2,3,4,5,6]
marks=[35,40,55,65,75,90]

plt.scatter(hours,marks)
plt.title("Study hours vs Marks")
plt.xlabel("Study Hours") ; plt.ylabel("Study Marks")

plt.show()

```

```python
import matplotlib.pyplot as plt 

height = [150, 155, 160, 165, 170, 175]
weight = [45, 48, 52, 58, 65, 70]

plt.scatter(height,weight,color="green",alpha=0.9,marker='x')
plt.title("Height vs Weight")
plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.show()

```

```python
import matplotlib.pyplot as plt 

# Histogram : A histogram Shows the freq dist of numerical data by grouping values into intervals called bins .

marks=[i for i in range(45,94,5)]

plt.hist(marks,bins=5,edgecolor="black")
plt.title("Dist of Student marks")

plt.xlabel("Marks")
plt.ylabel('Freq')

plt.show()
# bins controls how many intervals are used. Changing the number or width of bins can change the appearance of a histogram.
```
