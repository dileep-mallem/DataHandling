```python
# Seaborn : High Level , Statistical plots in one line 
# It can integrate pandas excellent and had beautiful aesthetics 
# Built on top of mat -> return a obj's
# Focus on statistical plots (dist,regression , categorical)

# Evry Seaor plot returns a Matplotlib Axes obj(or Facetgrid) .
# Seaborn for quick statistical plots, Matplotlib for custom control.
```

```python
# Praparing a pandas dataset 

import pandas as pd 
import numpy as np 
import seaborn as sns  
import matplotlib.pyplot as plt 

plt.style.use("dark_background")

np.random.seed(42)

df=pd.DataFrame({
    "StudyHours" : np.random.uniform(1,8,100),
    "Marks":np.clip(
        35 + np.random.uniform(1,8,100)*7 
        + np.random.normal(0,8,100) , 
        0,100
    ),
    "Department" : np.random.choice(
        ["AIML","CSE","ECE"],100
    ),
    "Gender" : np.random.choice(
        ["Male","Female"],100
    )
})

print(df.head())

```

```python
# Distribution Plots -> Helps u to unserstand freq,density , varibility and outliers in numerical data
# histplot() - Histogram
# shows how many observations falls into each interval 

sns.histplot(data=df,x="Marks",bins=10,kde=True) # kde=True overlays a smoothed density curve.

plt.title("Distribution of Marks")
plt.xlabel("Marks");plt.ylabel("Frequency")

plt.show()
```

```python
# kdeplot -> density curve 
# Kernel Density Estimation(KDe) estimates the prob density of numerical  data 

sns.kdeplot(data=df,x="Marks",fill=True)

plt.title("Density of Student Matks")
plt.xlabel("Marks");plt.ylabel("Estimated Density")

plt.show()

# Use KDE to observe the shape of a distribution. Its curve is an estimate, not a count of students.
```

```python
# boxplot() - box and whisker plot 
# Displays the medisn , quartiles , spread and potential outlier 

sns.boxplot(data=df,x="Department",y="Marks")

plt.title("Marks by Dept")
plt.xlabel("Dept");plt.ylabel("Marks")

plt.show()

# Middle line — median.
# Box — first quartile (Q1) to third quartile (Q3).
# Box height — interquartile range (IQR).
# Whiskers — extend to the most extreme observations within the usual \(1.5 \times IQR\) rule.
# Points beyond whiskers — potential outliers, not necessarily errors.
```

```python
# violinplot() - combines a box-like summary with a density shape 
#                to show where observations are concentrated 

sns.violinplot(
    data=df,
    x="Department",
    y="Marks",
    inner="quartile",
    cut=0
)

plt.title("Marks Distribution by Department")

plt.show()
# Boxplot vs violin plot:
#  A boxplot summarizes quartiles and outliers.
# A violin plot additionally shows the estimated distribution shape.
```

```python
# Realtional plots - shows how two or numerical variables are related 

# Scatterplot()

plt.style.use("classic")
sns.scatterplot(data=df,x="StudyHours",y="Marks",hue="Department",style="Gender",s=100)

plt.title("Study Hours vs Marks")
plt.show()

# hue — assigns colors according to a category.
# style — changes marker styles by category.
# s — controls marker size.
```

```python
# lineplot() - Useful for displaying trends, such as average marks across study-hour intervals.

sns.lineplot(
    data=df,
    x="StudyHours",
    y="Marks",
    hue="Department",
    errorbar=None
)

plt.title("Study Hours vs Marks by Dept")
plt.show()

# Seaborn's lineplot() aggregates repeated x-values by default, typically using a mean
# For raw observations, use plt.plot() or disable aggregation with estimator=None
```

```python
# categorical plots  - ompare numerical measurements across groups.

# Barplot() - compare group averages 

sns.barplot(
    data=df,
    x="Department",
    y="Marks",
    errorbar=None,
    color= "cyan",
    width=0.5
)
#
# A bar represents the estimated mean by default.
#  With errorbar=None, uncertainty intervals are hidden.
plt.title("Avg marks by Dept")
plt.show()
```

```python
# countplot() - count observations 

sns.countplot(
    data=df,
    x="Department",
    hue="Department",
    legend=False,
    width=0.5
)

plt.title("Number of Students by Dept")
plt.ylabel("Student Count")
plt.show()
```

```python
# stripplot() - didplay indivisual observations 

sns.stripplot(
    data=df,
    x="Department",
    y="Marks",
    jitter=True,
    alpha=0.7
)

plt.title("Individual Marks by Department")
plt.show()
# useful when you want to see individual data points rather than only summary statistics.
```

```python
# Heatmap - rep values using colors , It is often used to visualize correlation Matrix 

numeric_df=df[["StudyHours","Marks"]]

corr=numeric_df.corr() # df.corr() - Compute pairwise correlation of columns, excluding NA/null values.

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0
)

plt.title("Correlation Heatmap")

plt.show()
# annot=True — displays the numerical values inside the cells.
# cmap — selects the color scheme.
# vmin=-1, vmax=1 — set the correlation scale.
# - center=0 — centers the colormap at zero.
```

```python
# Pairplot - displays pairwise relationships between multiple numerical variables with distributions on the diagonal 
# Pairplots are useful during exploratory data analysis (EDA) to identify trends, clusters and possible outliers.
sns.pairplot(
    data=df,
    hue="Department",
    vars=["StudyHours","Marks"],
    diag_kind="hist"
)

plt.show()
# vars — selects numerical columns.
# diag_kind="hist" — shows histograms on the diagonal.
```
